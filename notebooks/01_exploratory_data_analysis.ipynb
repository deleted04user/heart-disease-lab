# Heart Disease Lab — reproducible EDA
This notebook reads the actual local UCI dataset and calls the same EDA workflow as `python scripts/eda.py`. Run the download command first. Full-dataset EDA is descriptive; preprocessing is fitted only inside training folds. Educational use only. Outputs are intentionally cleared to avoid committing source records.

```python
from pathlib import Path
import sys
root = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(root / 'src'))
from heart_disease.data import load_dataset
from heart_disease.eda import run_eda
dataset = load_dataset(root / 'data/raw/processed.cleveland.data')
dataset.metadata
```

```python
dataset.frame.info()
display(dataset.frame.head())
display(dataset.frame.describe().T)
display(dataset.frame.isna().sum().to_frame('missing'))
display(dataset.frame['target'].value_counts().sort_index())
```

```python
summary = run_eda(dataset, root / 'reports')
from IPython.display import Image, Markdown, display
display(Markdown((root / 'reports/eda_report.md').read_text(encoding='utf-8')))
for name in ['target_distribution', 'missing_values', 'numeric_distributions', 'numeric_by_target', 'correlations', 'categorical_distributions']:
    display(Image(filename=str(root / f'reports/figures/{name}.png')))
```

Read the computed report above. Historical hospital sampling, small sample size, missing observations, site selection and limited demographic coding constrain generalization.
